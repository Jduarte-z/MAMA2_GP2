```python
import os
import sys
import numpy as np
import pandas as pd
import time
from scipy.stats import chi2
import seaborn as sns


```

```python
print(f"working dir: {os.getcwd()}")
PCs = pd.read_csv(
    'pcs_MAMA_2026.tsv',
    sep=" ",
    header=None,
    names=['FILE','PC1','PC2','PC3','PC4','PC5','PC6']
)
PCs['Datasets'] = PCs['FILE'].str.replace(
    r'(\w+)\..+',  # match "data/<name>.<anything>"
    r'\1',              # keep just <name>
    regex=True
)

PCs



```

```python
PCs.loc[PCs['Datasets'].isin(['gp2EurFirstPass']),'SUPERPOP']='EUR'
PCs.loc[PCs['Datasets'].isin(['EASFirstPass']),'SUPERPOP']='EAS'
PCs.loc[PCs['Datasets'].isin(['P2B1-10saigeFirstPass','P1saigeFirstPass']),'SUPERPOP']='AMR'
PCs.loc[PCs['Datasets'].isin(['africaFirstPass']),'SUPERPOP']='AFR'
PCs.loc[PCs['Datasets'].isin(['SAFirstPass']),'SUPERPOP']='SOUTHAFR'
PCs.loc[PCs['Datasets'].isin(['Kishore_india_FirstPass']),'SUPERPOP']='SAS'
```
